# 02 - Exploratory Data Analysis (EDA)
Menganalisis tren inflasi Indonesia dan hubungannya dengan variabel makro lainnya.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_theme(style="whitegrid")
```

## 1. Load Data
Kita akan memuat dataset master yang sudah dibersihkan.

```python
df_path = '../data/processed/master_dataset.csv'
if os.path.exists(df_path):
    df = pd.read_csv(df_path, parse_dates=['date'])
    # Ensure target column
    if 'pcpipch' in df.columns and 'inflation_yoy' not in df.columns:
        df['inflation_yoy'] = df['pcpipch']
    display(df.head())
else:
    print("❌ Master dataset not found. Please run 01_scraping.ipynb first.")
    df = pd.DataFrame()
```

## 2. Visualisasi Tren Inflasi

```python
if not df.empty:
    plt.figure(figsize=(12, 6))
    sns.lineplot(data=df, x='date', y='inflation_yoy', marker='o')
    plt.title("Indonesia Annual Inflation (YoY) - Historical Trend")
    plt.ylabel("Inflation (%)")
    plt.xlabel("Year")
    plt.show()
else:
    print("⚠️ Skipping visualization: No data.")
```

## 3. Seasonality Analysis (Monthly)

```python
if not df.empty:
    df['month'] = df['date'].dt.month
    plt.figure(figsize=(10, 5))
    sns.boxplot(data=df, x='month', y='inflation_yoy')
    plt.title("Inflation Distribution by Month (Seasonality Check)")
    plt.show()
```

## 4. Correlation Matrix

```python
if not df.empty:
    corr = df.corr()
    plt.figure(figsize=(10, 8))
    sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
    plt.title("Correlation Matrix of Economic Indicators")
    plt.show()
```
